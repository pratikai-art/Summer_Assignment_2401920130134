```python
s1 = []
s2 = []

def push(x):
    s1.append(x)

def pop():

    if not s2:
        while s1:
            s2.append(s1.pop())

    return s2.pop()

def peek():

    if not s2:
        while s1:
            s2.append(s1.pop())

    return s2[-1]

def empty():

    return len(s1) == 0 and len(s2) == 0
```

```python
push(1)
push(2)
push(3)

print(peek())   
print(pop())    
print(pop())  
print(empty())  
print(pop())    
print(empty()) 
```

```python
q = []

def ping(t):

    q.append(t)

    while q[0] < t - 3000:
        q.pop(0)

    return len(q)

print(ping(3001))
print(ping(3002))
```

```python
from collections import deque

def maxSlidingWindow(nums, k):

    q = deque()
    ans = []

    for i in range(len(nums)):

        while q and q[0] <= i - k:
            q.popleft()

        while q and nums[q[-1]] < nums[i]:
            q.pop()

        q.append(i)

        if i >= k - 1:
            ans.append(nums[q[0]])

    return ans

nums = [1,3,-1,-3,5,3,6,7]
k = 3 
print(maxSlidingWindow(nums, k))
```
