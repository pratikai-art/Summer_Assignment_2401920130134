```python
class TreeNode:
    def __init__(self,val):
        self.val=val
        self.left=None
        self.right=None

root1=TreeNode(50)

root1.left=TreeNode(40)
root1.right=TreeNode(60)


root1.left.left=TreeNode(30)
root1.left.right=TreeNode(20)

root1.right.left=TreeNode(10)
root1.right.right=TreeNode(5)


root2=TreeNode(50)

root2.left=TreeNode(40)
root2.right=TreeNode(60)


root2.left.left=TreeNode(30)
root2.left.right=TreeNode(20)

root2.right.left=TreeNode(10)
root2.right.right=TreeNode(5)

def sameT(p,q):
    if  p is None and q is None:    #to check similarty 
        return True

    if p is None or q is None:       #to check diifernnce
        return False

    if p.val !=q.val:
        return False
        
    left=sameT(p.left,q.left) 
   
    right=sameT(p.right,q.right)
    

    return left and right

print(sameT(root1,root2))    
```
