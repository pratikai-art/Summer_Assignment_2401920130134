```python
def isAna(s,t):
    feq_s={}
    feq_t={}
    for ch in s:
        feq_s[ch]=feq_s.get(ch,0)+1
    for ch in t:
        feq_t[ch]=feq_t.get(ch,0)+1

    if(feq_s==feq_t):
        return True
    else:
        return False
```

```python
print(isAna("anagram","nagaram"))
print(isAna("rat","car"))
```

```python
from collections import Counter
def isAna(s,t):
    
    if(Counter(s)==Counter(t)):
        return True
    else:
        return False

print(isAna("anagram","nagaram"))
print(isAna("rat","car"))
```
