```python
#lps

def build_lps(pattern):
    lps=[0]*len(pattern)
    lenght=0
    i=1

    while i<len(pattern):
        if(pattern[i]==pattern[lenght]):
            lenght+=1
            lps[i]=lenght
            i+=1
        else:
            if(lenght !=0):
                lenght=lps[lenght-1]

            else:
                lps[i]=0
                i+=1
    return lps


print(build_lps("abcabcabcabc"))  
print(build_lps("aba"))  
```

```python
def kmp_search(text,pattern):
    lps=build_lps(pattern)
    i=0
    j=0
    while i<len(text):
        if pattern[j]==text[i]:
            i+=1
            j+=1
            if j==len(pattern):
                return i-j
        else:
            if (j!=0):
                j=lps[j-1]
            else:
                i+=1

    return -1           
    
```

```python
print(kmp_search("sadbutsad","sad"))
print(kmp_search("leetcode","leeto"))
```

```python
def strStr(haystack,needle):
    left=0
    right=len(needle)

    while left<len(haystack)-right:
        if (needle==haystack[left:right]):
            return left
        else:
            left+=1
            right+=1
    return -1

print(strStr("sadbutsad","sad"))
print(strStr("leetcode","leeto"))
```

```python
from collections import Counter
def isSubsequence(s,t):
    feq_s=Counter(s)
    feq_t=Counter(t)
    if(feq_s<=feq_t):
        return True
    else:
        return False

print(isSubsequence("abc","ahbgdc"))      
print(isSubsequence("axc","ahbgdc")) 
```
