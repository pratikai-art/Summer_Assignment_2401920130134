```python
def reversW(s):
    stack = []
    result = []

    for ch in s:
        if ch == " ":
            while stack:
                result.append(stack.pop())
            result.append(" ")
        else:
            stack.append(ch)

    while stack:
        result.append(stack.pop())

    return "".join(result)

print(reversW("Let's take LeetCode contest"))    
print(reversW("Mr Ding"))  
 
```
