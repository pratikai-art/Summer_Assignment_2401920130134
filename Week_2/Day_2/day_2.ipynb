```python
def dupli(s):
    list=[]
    for ch in s:
        if(ch in list):
            return True
        else:
            list.append(ch)
    
print(dupli(""))
```

```python
def duplicate(s):
    left = 0
    seen = set()
    max_len = 0

    for right in range(len(s)):

        while s[right] in seen:
            seen.remove(s[left])
            left += 1

        seen.add(s[right])

        max_len = max(max_len, right - left + 1)

    return max_len

print(duplicate("abcabcbb"))
```

```python
from collections import Counter

def checki(s1,s2):
    feq_r=Counter(s1)
    n=len(s1)
    
    
    left=0
    right= n
    
    

    while left <= len(s2)-n:
        q=Counter(s2[left:right])
        
        if feq_r==q:
            return True
        
        left+=1
        right+=1
    return False

    
print(checki("ab","eidbaooo")) 
print(checki("ab","eidboaoo")) 
        
```

```python
from collections import Counter
def findAnagrams(s,p):
    feq_q=Counter(p)
    left=0
    right=len(p)
    list_p=[]
    while left<=len(s)-len(p):
        q=Counter(s[left:right])

        if feq_q==q:
            list_p.append(left)
            
        left+=1
        right+=1
    return list_p    

print(findAnagrams("abab","ab"))
print(findAnagrams("cbaebabacd","abc"))

    
```
